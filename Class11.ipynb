```python
#flatmap -> 1차원으로 변경 , split 단어로 나뉨
```

```python
myRdd2\
    .flatMap(lambda x:x.split())\
    .map(lambda x:(x,1))\
    .groupByKey()\
    .mapValues(sum)\
    .take(10)
```

```python
def f(x): return len(x)

myRdd2\
    .flatMap(lambda x:x.split())\
    .map(lambda x:(x,1))\
    .groupByKey()\
    .mapValues(f)\
    .sortByKey(True)\
    .take(10)
```

```python
wc=myRdd2\
    .flatMap(lambda x:x.split())\
    .map(lambda x:(x,1))\
    .groupByKey()\
    .mapValues(sum)\
    .sortByKey(True)\
    .take(10)

for k,v in wc:
    print (f"단어:{k}\t\t빈도:{v}")
```

```python
import os

stopwords = set(['및','이를','등','이','이런','그와','또는','두', '이와', '전', '간'])
d = dict()

with open(os.path.join("data", "ds_bigdata_wiki.txt"), 'r', encoding='utf-8') as f:
    for sent in f.readlines(): 
        _words = sent.split()  # split into words  split은 리스트로 그릇에 담긴다
        for word in _words:
            if word not in stopwords: # remove stopwords
                if word not in d:
                    d[word]=1
                else:
                    d[word]=d[word]+1 #있으면 +1을 한다 -> d[k]를 읽어오곤 한다
```

```python
dSorted = {k: v for k, v in sorted(d.items(), key=lambda x: x[1], reverse=True)}
```

```python
d1 = dict()
for key, value in dSorted.items():
    if value>5:
        d1[key]=value
        print (f"{key}\t{value}")
```

```python

%%writefile src/ds_spark_rdd_hello.py
#!/usr/bin/env python3
# -*- coding: UTF-8 -*- 
import pyspark
import os

os.environ['HADOOP_HOME']=os.getcwd()
os.environ["PATH"] += os.path.join(os.environ['HADOOP_HOME'], 'bin')

def doIt():
    print ("---------RESULT-----------")
    print (spark.version)
    spark.conf.set("spark.logConf","false")
    rdd=spark.sparkContext.parallelize(range(1000), 10)
    print ("mean=",rdd.mean())
    nums = spark.sparkContext.parallelize([1, 2, 3, 4])
    squared = nums.map(lambda x: x * x).collect()
    for num in squared:
        print ("{} ".format(num))

if __name__ == "__main__":

    myConf=pyspark.SparkConf()
    spark = pyspark.sql.SparkSession.builder\
        .master("local")\
        .appName("myApp")\
        .config(conf=myConf)\
        .getOrCreate()
    doIt()
    spark.stop()
```
