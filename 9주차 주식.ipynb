```python
import pandas as pd
```

```python
_myDf = pd.DataFrame(_ohlcv, index=pd.date_range('2024-10-01', periods=17))
```

```python
_myDf
```

```python
_myDf['Close-1'] = _myDf['Close'].shift(-1)
_myDf['Target'] = (_myDf['Close'].shift(-1) > _myDf['Close']).astype(int)

#shift는 한 줄 내릴때 사용하는 건데 여기서는 -1이니까 한 줄 올린 거얌
```

```python
#data를 보고 상승했는지 하강했는지 Target 변수를 이용하여 이분화 가능
```

```python
features = ['SMA_10', 'SMA_20']
X = _myDf[features].iloc[:-1]  # [:-1] --> 마지막 줄 제외, features만 선별 (Close 제외)
y = _myDf['Target'].iloc[:-1]  # [:-1] --> 마지막 줄 제외
```

```python
import sklearn
from sklearn.model_selection import train_test_split
#학습시키는 방법

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier

#randomforest를 이용한 분류
model = RandomForestClassifier(n_estimators=100, random_state=42)
```

```python
#모델 학습시키기

model.fit(X_train, y_train)
```

```python
#모델에 test data 넣어보기

y_pred = model.predict(X_test)
```

```python
X_test['Prediction'] = y_pred

#test data로 값 예측하기
```
